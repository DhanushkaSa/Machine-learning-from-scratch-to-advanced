```python
class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = ["laptop", "mouse", "pen drive"]

    def __len__(self):
        return len(self.items)

    def __getitem__(self, index):
        return self.items[index]


my_cart = ShoppingCart("Dhanushka")
print(len(my_cart))
print(my_cart[1])
```

call


```python
class Scalar:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, x):
        return x * self.factor


my_scalar = Scalar(2)
print(my_scalar(10))
```

repr


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


person_1 = Person("Dhanushka", 23)
print(person_1)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f'Person(name = {self.name}, age = {self.age})'

    def __str__(self):
        return "This is the person class"


person_1 = Person("Dhanushka", 23)
print(person_1)
print(repr(person_1))
```

# Vectors


```python
import math


def distance_between_vectors(a, b):
    value = 0

    if len(a) == len(b):
        for i in range(len(a)):
            value += (a[i]-b[i])**2

        return math.sqrt(value)
    else:
        return "Vectors are not of same length"


print(distance_between_vectors([1, -1, 3], [3, 2, -2]))
```

```python
def calculate_manhattan_distance(a,b):
    return sum([ abs(a[i]-b[i]) for i in range(len(a)) ])

 
print(calculate_manhattan_distance([-1,3], [-4,-1]))
```

```python
def calculate_dotproduct(a,b):
    if len(a)==len(b):
        return sum([a[i]*b[i] for i in range(len(a))])
    
print(calculate_dotproduct([-3,4,-1],[8,0,6]))
```
