```python
class ShoppingCart:
    def __init__(self,customer):
        self.customer = customer
        self.items = ["A","B","C"]
        self.item=[1,2,3,4,5]

    def __len__(self):
        return len(self.items)
    
    
    
my_cart = ShoppingCart("Dhanushka")
print(len(my_cart))
```

```python
class ShoppingCart:
    def __init__(self,customer):
        self.customer = customer
        self.items = ["A","B","C"]
        self.item = [1,2,3,4]
        

    def __getitem__(self, index):
        return self.items[index], self.item[index]
    
    
    
    
    
my_cart = ShoppingCart("Dhanushka")
print(my_cart[1])
```

```python
class Scaler:
    def __init__(self,factor):
        self.factor = factor

    def __call__(self,x):
        return x * self.factor
    
first = Scaler(1)
print(first(20))
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __repr__(self):
        return f'Person(name: {self.name}, age: {self.age})'
    
    def __str__(self):
        return "This is the person class"
    

person_1=Person("Kasun",23)
print(person_1)
print(repr(person_1))
```

# Vectors

```python
import math

def calculate_distance_between_vectors(v1,v2):
    if len(v1)!=len(v2):
        return "vectors must be of same length"

    return math.sqrt(sum([(v1[i]-v2[i])**2 for i in range(len(v1))]))


print(calculate_distance_between_vectors([1, -1, 3], [3, 2, -2]))
```

```python
def calculate_manhattan_distance(v1,v2):
    if len(v1)!=len(v2):
        return "vectors must be of same length"

    return sum([abs(v1[i]-v2[i]) for i in range(len(v1))])

print(calculate_manhattan_distance([-1,3], [-4,-1]))
```
