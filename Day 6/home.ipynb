```python
class ShoppingCart:
    def __init__(self,customer):
        self.customer = customer
        self.items = ["A","B","C"]
        self.item=[1,2,3,4,5]

    def __len__(self):
        return len(self.items)
    
    
    
my_cart = ShoppingCart("Dhanushka")
print(len(my_cart))
```

```python
class ShoppingCart:
    def __init__(self,customer):
        self.customer = customer
        self.items = ["A","B","C"]
        self.item = [1,2,3,4]
        

    def __getitem__(self, index):
        return self.items[index], self.item[index]
    
    
    
    
    
my_cart = ShoppingCart("Dhanushka")
print(my_cart[1])
```
