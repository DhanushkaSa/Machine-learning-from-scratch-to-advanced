# Vectors


```python
import math


def calculate_distance_between_vectors(v1, v2):
    if len(v1) != len(v2):
        return "vectors must be of same length"

    return math.sqrt(sum([(v1[i]-v2[i])**2 for i in range(len(v1))]))


print(calculate_distance_between_vectors([1, -1, 3], [3, 2, -2]))
```

```python
def calculate_manhattan_distance(v1, v2):
    if len(v1) != len(v2):
        return "vectors must be of same length"

    return sum([abs(v1[i]-v2[i]) for i in range(len(v1))])


print(calculate_manhattan_distance([-1, 3], [-4, -1]))
```

```python
def calculate_dot_product(v1, v2):
    if len(v1) != len(v2):
        return "vectors must be of same length"

    return sum([v1[i]*v2[i] for i in range(len(v1))])


print(calculate_dot_product([-3, 4, -1], [8, 0, 6]))
```

```python
def euclidean_distance(a, b):
    squared_sum = 0
    for i in range(len(a)):
        squared_sum += (a[i]-b[i])**2
    return math.sqrt(squared_sum)


student_a = [80, 70, 90]
student_b = [82, 68, 88]

print(euclidean_distance(student_a, student_b))
```

Manhattan distance


```python
def manhattan_distance(a, b):
    total = 0
    for i in range(len(a)):
        total += abs(a[i]-b[i])
    return total


print(manhattan_distance([1, 2], [4, 6]))
```

```python
def f1_score(precision: float, recall: float) -> float:
    """Return the harmonic mean of precision and recall."""
    if precision+recall == 0:
        return 0.0
    return 2*(precision*recall)/(precision+recall)


print(f1_score(0.8, 0.6))
```

```python
class RunningAverage:
    def __init__(self):
        self.total = 0
        self.count = 0

    def add(self,value):
        self.total += value
        self.count += 1

    def value(self):
        if self.count == 0:
            return 0.0
        return self.total / self.count
    
loss_tracker = RunningAverage()
for loss in [0.9,0.7,0.5]:
    loss_tracker.add(loss)
print(loss_tracker.value())
```

```python
class BaseModel:
    def __init__(self,name):
        self.name = name

    def predict(self,x):
        raise NotImplementedError('Subclass must implement predict')
    
    def describe(self):
        return f"Model : {self.name}"
    
class ThreshouldModel(BaseModel):
    def __init__(self,name,threshold):
        super().__init__(name)
        self.threshold = threshold

    def predict(self,x):
        return 1 if x >= self.threshold else 0
    

model = ThreshouldModel('baseline',0.5)
print(model.describe())  # baseline
print(model.predict(0.8))  # 1
print(model.predict(0.3)) # 0
```
