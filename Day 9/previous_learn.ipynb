```python
M = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(M[0, 0])  # 10
print("-"*30)
print(M[1])    # [40, 50, 60]
print("-"*30)
print(M[:, 0])  # 10,40,70
print("-"*30)
print(M[0:2, 1:3])  # [[20,30],[50,60]]
```

```python
scores = np.array([0.2, 0.9, 0.5, 0.95, 0.1])

mask = scores >= 0.5
print(mask)

print(scores[mask])
print("-"*30)

print(scores[scores >= 0.5])
```
