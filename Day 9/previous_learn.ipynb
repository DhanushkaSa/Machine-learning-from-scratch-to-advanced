```python
M = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(M[0, 0])  # 10
print("-"*30)
print(M[1])    # [40, 50, 60]
print("-"*30)
print(M[:, 0])  # 10,40,70
print("-"*30)
print(M[0:2, 1:3])  # [[20,30],[50,60]]
```
