Data Clean (Remove null values)


```python
df = pd.read_csv("customers_tiny.csv")
print(df)
```

```python
clean_df = df.dropna()
print(clean_df)
```

```python
print(df.isna())
```

```python
print(df.isna().sum())
```

```python
print(df)
```

```python
df["age"] = df["age"].fillna(df["age"].mean())
df["score"] = df["score"].fillna(df["score"].median())

print(df)
```

```python
print(df["age"].dtype)
```

Convert data type for a entire column


```python
df["age"] = df["age"].astype("int64")
print(df)
```

Sort values


```python
print(df.sort_values("score"))
```

```python
print(df.sort_values("score", ascending=False))
```
