```python
import numpy as np
```

```python
list_a = [1,2,3]
list_b = [4,5,6]

print(list_a + list_b)

a = np.array([1,2,3])
b = np.array([4,5,6])

print(a + b)
```

```python
from_list = np.array([1,2,3])
matrix = np.array([[1,2,3],[4,5,6]])

print(from_list)
print("-"*30)
print(matrix)
```

```python
print(np.zeros((2,3)))
print("-"*30)
print(np.ones((2,2)))
print("-"*30)
print(np.arange(0,10,2))
print("-"*30)
print(np.linspace(0,1,5))
print("-"*30)
print(np.eye(3))
```

```python
x = np.array([[1,2,3],[4,5,6]],dtype="int16")

print(x.shape)
print(x.ndim)
print(x.size)
print(x.dtype)
print(len(x))
```

```python
print(np.array([1,2,3]).dtype)
print(np.array([1.0,2,3]).dtype)

# Convert dtype with astype
prices = np.array([1.5,2.5,3.5])
as_int = prices.astype(int)
print(as_int)
```
