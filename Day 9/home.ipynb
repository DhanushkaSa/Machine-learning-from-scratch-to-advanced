```python
import pandas as pd
```

Pandas Series

```python
ages = pd.Series([22,35,28,41])
print(ages)
print(ages.shape)
```

```python
ages = pd.Series([22,35,28,41],index=["Dhanushka","Sandun","Pathum","Kasun"])
print(ages)
```

```python
print(ages.loc["Pathum"])
```

```python
ages = pd.Series([25,30,48,75,65])
print(ages.loc[2])
print(ages.shape)
```

Pandas DataFrame

```python
df = pd.DataFrame({
    "name":["Dhanushka","Sandun","Kasun","Pathum","Sunil"],
    "age":[25,54,34,62,47],
    "score":[0.64,0.47,0.58,0.94,0.75]

})

print(df)
print(type(df))
```

Label base accessing

```python
print(df.loc[2,"age"])
```

Index base accessing

```python
print(df.iloc[2,1])
```
