# Visualization Basics

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    'age': [22,23,28,41,33,26,45,31],
    'score': [0.81,0.44,0.92,0.67,0.70,0.55,0.60,0.88],
    'city': ['Colombo','Kandy','Galle','Colombo','Galle','Kandy','Colombo','Galle']
})

```

```python
# Histogram

plt.figure()   # Define Canvas
plt.hist(df['age'],bins=5,edgecolor='white')  # Consider age column
plt.title('Age Distribution')  # Title of the plot
plt.xlabel("age")  # X-axis label
plt.ylabel("count")  # Y-axis label
plt.show()  # Show the plot
```

```python
city_mean = df.groupby('city')['score'].mean()
city_mean
```

Bar Chart

```python
plt.figure()
city_mean.plot(kind='bar', color='orange', edgecolor='black')
plt.title('Average Score by City')
plt.xlabel('City')
plt.ylabel('Average Score')
plt.show()
```

Scatter Plot

```python
plt.figure()
plt.scatter(df["age"],df["score"])
plt.title('Age vs Score')
plt.xlabel('Age')
plt.ylabel('Score')
plt.show()
```

Box plot

```python
plt.figure()
sns.boxplot(y=df['score'], x=df['city'])
plt.title("score spread by city")
plt.show()
```

Correlation heatmap

```python
numeric = df[['age','score']]
corr_matrix = numeric.corr()
print(corr_matrix)

plt.figure()
sns.heatmap(corr_matrix,annot=True,cmap="coolwarm",vmin=-1,vmax=1)
plt.title("Correlation heatmap")
plt.show()
```

# Sigmoid Function

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = 5
print(sigmoid(x)) 
```
