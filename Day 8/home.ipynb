Boolean Masking


```python
scores = np.array([0.6, 0.9, 0.3, 0.2, 0.75, 0.1])

temp_arr = scores > 0.5

print(temp_arr)

f_arr = scores[temp_arr]
print(f_arr)

print("-"*20)

print(scores[scores > 0.6])
```
