```python
scores = np.array([0.6, 0.9, 0.3, 0.2, 0.75, 0.1])

temp_arr = scores > 0.5

print(temp_arr)

f_arr = scores[temp_arr]
print(f_arr)

print("-"*20)

print(scores[scores > 0.6])
```

# Linear algebra basics


Dot Product


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a.dot(b))
print("-"*30)
print(a @ b)
```

Find Magnitude


```python
vector_a = np.array([3, 2, 1])
print(np.linalg.norm(vector_a))
```

Cosine Similarity


```python
vector_a = np.array([3, 4, 5])
vector_b = np.array([2, 4, 8])


def cosine_similarity(vector_a, vector_b):
    return (vector_a @ vector_b)/(np.linalg.norm(vector_a)*(np.linalg.norm(vector_b)))


print(cosine_similarity(vector_a, vector_b))
```

The bridge: X @ w + b


```python
X = np.array([
    [0.8, 0.4, 0.9],
    [0.2, 0.7, 0.6]
])

w = np.array([0.5, -0.2, 0.3])
b = 0.1

scores = X @ w + b
print(scores)
```

Matrix Vector Multiplication


```python
X = np.array([
    [0.8, 0.4, 0.9],
    [0.2, 0.7, 0.6]
])

w = np.array([
    [0.5],
    [-0.2],
    [0.3]
]
)
b = 0.1

scores = X @ w + b
print(X.shape)
print(w.shape)

print(scores)
print(scores.shape)
```

Matrix - Matrix multiplication


```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

C = A @ B
print(C.shape)
print("-"*30)
print(C)
```

```python
X = np.array([
    [0.1, 0.2, 0.3],
    [0.4, 0.1, 0.5],
    [0.2, 0.8, 0.1],
    [0.6, 0.3, 0.2]
])

W = np.array([
    [0.5, -0.2],
    [0.1, 0.4],
    [0.3, 0.6]
])

b = np.array([0.05, -0.1])

Y = X @ W + b

print(Y.shape)
print(Y)
```
