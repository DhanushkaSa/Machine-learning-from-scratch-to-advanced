Boolean Masking


```python
scores = np.array([0.6, 0.9, 0.3, 0.2, 0.75, 0.1])

temp_arr = scores > 0.5
print(temp_arr)

filtered_arr = scores[temp_arr]
print(filtered_arr)
```

```python
print(scores[scores > 0.5])
```

```python
scores = np.array([
    [3, 6, 4, 0.4],
    [1, 5, 8, 0.6],
    [2, 9, 12, 0.8],
    [4, 3, 7, 0.3]
])

print(scores[scores[:, -1] > 0.5])
```

Copy vs View


```python
base = np.array([6, 8, 5, 3, 4])
view = base[:3]
print(base)
print(view)

view[0] = 100
print(view)
print(base)
```

```python
base = np.array([6, 8, 5, 3, 4])
view = base[:3].copy()
print(base)
print(view)

view[0] = 100
print(view)
print(base)
```

# Linear algebra basics


Dot product


```python
vector_1 = np.array([3, -1, 2])
vector_2 = np.array([2, 6, 1])

dot_product = vector_1 @ vector_2
print(dot_product)


# Don't use this method
print(np.dot(vector_1, vector_2))
```

```python
vector_a = np.array([1, 2, 2])

print(np.linalg.norm(vector_a))
```

```python
vector_a = np.array([3, 4, 5])
vector_b = np.array([1, 2, 3])

cosine_similarity = vector_a @ vector_b / \
    (np.linalg.norm(vector_a) * np.linalg.norm(vector_b))
print(cosine_similarity)
```

Reshaping & Flattening


```python
arr_1 = np.array([1, 2, 3, 4, 5, 6])
print(arr_1.shape)

arr_2 = arr_1.reshape(2, 3)
print(arr_2)
print(arr_2.shape)


print(arr_1.reshape(2, 4))
```

```python
arr_1 = np.array([1, 2, 3, 4, 5, 6])
arr_2 = arr_1.reshape(3, -1)
print(arr_2)
```

```python
arr_1 = np.array([
    [1, 2, 3],
    [10, 20, 30],
    [100, 200, 300]
])

print(arr_1.flatten())
print(arr_1.reshape(1, -1))
```

Stacking and Concatenating arrays


```python
a = np.array([1, 5, 7])
b = np.array([4, 3, 10])

c = np.concatenate((a, b), axis=0)
print(c)
```

```python
a = np.array([
    [2, 5, 7, 1],
    [9, 12, 11, 13],
    [7, 6, 18, 12],
])

b = np.array([
    [7, 32, 12, 11],
    [9, 6, 7, 12],
    [5, 10, 6, 9]

])

c = np.concatenate((a, b), axis=0)
print(c)
```

```python
d = np.concatenate((a, b), axis=1)
print(d)
```

```python
a = np.array([1, 5, 7])
b = np.array([4, 3, 10])

c = np.vstack([a, b])
print(c)

print(np.hstack([a, b]))
```

```python
a = np.array([
    [1, 5, 7],
    [2, 4, 8]
])

b = np.array([
    [4, 5, 6],
    [3, 7, 8]
])

c = np.hstack([a, b])
print(c)
```

Matrix-Vector multiplication


```python
x = np.array([
    [3, 2, 100],
    [1, 2, 60],
    [4, 3, 20],
    [2, 5, 120]
])

w = np.array([
    [0.3],
    [0.2],
    [0.1]
])

b = 0.8

y = x @ w+b
print(y)
```

Identity Matrix


```python
I = np.eye(3, dtype="int64")
print(I)

a = np.array([
    [1, 2, 3],
    [10, 20, 30],
    [100, 200, 300]
])
print("-"*30)
print(a @ I)
```
