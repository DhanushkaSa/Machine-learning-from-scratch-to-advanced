# Numpy


```python
import numpy as np
```

```python
array_1 = np.array([9, 10, 11, 12])
print(array_1)
print(type(array_1))
```

```python
array_2 = np.array((9, 10, 11, 12))
print(array_2)
print(type(array_2))
```

```python
list_1 = [1, 2, 3, 4]
list_2 = [2, 4, 6, 8]
print(list_1 + list_2)
```

```python
list_1 = np.array([1, 2, 3, 4])
list_2 = np.array([2, 4, 6, 8])
print(list_1 + list_2)
```

```python
print(list_1-list_2)
print(list_1*list_2)
print(list_1/list_2)
```

```python
def calculate_dot_product(a, b):
    if len(a) == len(b):

        return sum(a * b)


print(calculate_dot_product(np.array([1, 2, 3, 4]), np.array([2, 4, 6, 8])))
```

```python
array_1 = np.array([
    [1, 9, 5, 8],
    [4, 5, 3, 6],
    [6, 5, 4, 10]
])

print(array_1[1])  # array_1[1,:]
print("----------------")
print(array_1[0:3, 2])

print("----------------")
print(array_1[1:])
print("----------------")
print(array_1[0:2, 1::])
```

Scalar multiplication


```python
array_1 = np.array([
    [1, 9, 5, 8],
    [4, 5, 3, 6],
    [6, 5, 4, 10]
])

print(2*array_1)
```

```python
c = np.array([
    [1, 2],
    [3, 4]
])

d = np.array([
    [5, 6],
    [7, 8]
])

print(c*d)
```

```python
a = np.array([
    [1, -2, -6],
    [3, 4, -3],
    [3, 1, 2]
])

b = np.array([
    [1, 4, 3],
    [0, -2, 3],
    [6, 1, 0]
])

print(a@b)
print("--------------------------------------------------")
print(b@a)
print("--------------------------------------------------")

print(np.matmul(a, b))
print("--------------------------------------------------")
print(np.dot(a, b))
```
