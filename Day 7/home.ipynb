```python
def calculate_dot_product(v1, v2):
    if len(v1) == len(v2):
        return sum(v1*v2)


print(calculate_dot_product(np.array([2, 4, 6]), np.array([3, 5, 7])))
```

```python
a = np.array([
    [1, -2, -6],
    [3, 4, -3],
    [3, 1, 2]
])

b = np.array([
    [1, 4, 3],
    [0, -2, 3],
    [6, 1, 0]
])

print(a@b)
print("--------------------------------------------------")
print(b@a)
print("--------------------------------------------------")

print(np.matmul(a, b))
print("--------------------------------------------------")
print(np.dot(a, b))
```

Indexing & Slicing

```python
M = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print(M[0,0])
print(M[1])
print(M[:,0])
print(M[0:2,1:3])
```

Boolean masking

```python
scores = np.array([0.2,0.9,0.5,0.95,0.1])

mask = scores >= 0.5
print(mask)
print(scores[mask])

print(scores[scores >= 0.5])
```

```python
X = np.array([
    [1,2,0.7],
    [3,4,0.9],
    [5,6,0.3]
])

print(X[X[:,2]> 0.8])
```

```python
base = np.array([1,2,3,4])
view = base[0:2]
view[0] = 99
print(base)

base2 = np.array([1,2,3,4])
safe = base2[0:2].copy()
safe[0] = 99
print(base2)
```

Broadcasting

```python
X = np.array([
    [1,2,3],
    [4,5,6]
])

column_mean = X.mean(axis = 0)
print(column_mean)

centered = X - column_mean
print(centered)

```

```python
A = np.array([
    [1,2,3],
    [4,5,6]
])

print(A.sum())
print(A.mean())
print(A.max())
print(A.min())
```

```python
print(A.sum(axis = 0))
print(A.sum(axis = 1))
```
