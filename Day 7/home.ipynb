```python
list_a = [1, 2, 3]
list_b = [4, 5, 6]
print(list_a+list_b)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a+b)
```

```python
print(a-b)
print(a*b)
print(a/b)
```

```python
def calculate_dot_product(v1, v2):
    if len(v1) == len(v2):
        return sum(v1*v2)


print(calculate_dot_product(np.array([2, 4, 6]), np.array([3, 5, 7])))
```
