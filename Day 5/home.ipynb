```python
def f1_score(precision: float, recall: float) -> float:
    """
    Return the harmonic mean of precision and recall.

    """

    if precision + recall == 0:
        return 0.0
    return 2*(precision*recall)/(precision+recall)


print(f1_score(0.8, 0.6))
```

```python
class RunningAverage:
    def __init__(self):
        self.total = 0.0
        self.count = 0

    def add(self, value):
        self.total += value
        self.count += 1

    def value(self):
        if self.count == 0:
            return 0.0
        return self.total/self.count


loss_tracker = RunningAverage()
for loss in [0.9, 0.7, 0.5]:
    loss_tracker.add(loss)
print(loss_tracker.value())
```

```python
class BaseModel:
    def __init__(self, name):
        self.name = name

    def predict(self, x):
        raise NotImplementedError("Subclasses must implement predict")

    def describe(self):
        return f'Model : {self.name}'


class ThresholdModel(BaseModel):
    def __init__(self, name, threshold):
        super().__init__(name)
        self.threshold = threshold

    def predict(self, x):
        return 1 if x >= self.threshold else 0


model = ThresholdModel('baseline', 0.5)
print(model.describe())
print(model.predict(0.8))
print(model.predict(0.3))
```

```python
class Car:
    def __init__(self, brand, color):
        self._brand = brand
        self._color = color

    def _display(self):
        print(f'{self._brand}, {self._color} color car')


class CarSub(Car):

    def info(self):
        super()._display()
        print(f'{self._brand}, {self._color} color car')


car_1 = CarSub("BMW", "Red")
car_1.info()
```

```python
class Car:
    def __init__(self,brand,color):
        self._brand = brand
        self._color = color

    def display(self):
        print(f'Car brand: {self._brand}, Color: {self._color}')

    @property
    def get_brand(self):
        return self._brand
    

car_1 = Car("BMW","Red")
car_1.display()
print(car_1.get_brand)
```

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return (f'Engine with {self.horsepower} HP is roaring to life!')


class Battery:
    def __init__(self, capacity):
        self.capacity = capacity

    def charge(self):
        return (f'Battery charging to {self.capacity} kwh...')


# The car class USES Engine and Battery (Composition)

class Car:
    def __init__(self, model, horse_power, charging_capacity):
        self.horse_power = Engine(horse_power)
        self.charging_capacity = Battery(charging_capacity)
        self.model = model


car_1 = Car("Benz", 5000, 75)
print(car_1.horse_power.start())
print(car_1.charging_capacity.charge())
print(car_1.model)
```
