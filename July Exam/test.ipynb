### 4 Bullet-Point Findings for a Future ML Model

1. **Strong Predictor (`score`):** The `score` variable has a strong positive correlation ($r \approx 0.54$) with the purchase decision. Purchased customers have a much higher average score ($\approx 0.75$) than non-purchased customers ($\approx 0.53$).
2. **Class Imbalance:** The target `purchased` is imbalanced (approx. 31% purchase rate vs 69% non-purchase rate). The future model must optimize for metrics like F1-Score or ROC-AUC instead of simple accuracy.
3. **Non-Linear Demographics:** Key features like `age` ($r \approx 0.01$) and `income` ($r \approx 0.07$) show very low linear correlations. Tree-based models (e.g., XGBoost) or non-linear feature engineering are needed to capture their interactions.
4. **Geographic Variation:** Standarized city categories show distinct purchase behavior. Negombo has the highest purchase rate ($\approx 39.4\%$), whereas Jaffna has the lowest ($\approx 26.3\%$). Categorical encoding will capture these regional trends.
