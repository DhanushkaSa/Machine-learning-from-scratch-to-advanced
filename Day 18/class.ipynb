```python
import torch
import torch.nn as nn
```

```python
class CustomerPredictor(nn.Module):
  def __init__(self):
    super().__init__()
    self.hidden = nn.Linear(3,8) # Linear wx+b , input=3 hidden nodes = 8 (3,8)
    self.output = nn.Linear(8,1) # Linear wx+b , hidden nodes = 8 outputs = 1 (8,1)

  def forward(self,x):  # this forward function means neural network forward propagation , Module super class override this forward function
    x = torch.relu(self.hidden(x))  # relu is activation function , this is tensor (similar multidimensional numpy array. we want to convert our data tensor)
    return self.output(x)  # return only Z score not a probability

# Instantiate the model
model = CustomerPredictor()
```

```python
print(list(model.parameters()))

# 1st tensor = weights (Input -> hidden nodes)
# 2nd tensor = hidden nodes bias
# 3rd tensor = weights (hidden -> output)
# 4th tensor = output bias
```

```python
sample_tensor = torch.tensor([[0.5,1.0,2.0]])
print(sample_tensor)
print(sample_tensor.shape)
```
